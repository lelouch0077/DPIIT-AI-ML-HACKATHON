```python
import json
import pytesseract
from PIL import Image
!pip install langdetect

from langdetect import detect
image_path = "/kaggle/input/checktext/input_image.png"       
layout_json_path = "/kaggle/input/checktext/layout.json"      
output_json_path = "final_output.json"

image = Image.open(image_path)

with open(layout_json_path, "r", encoding="utf-8") as f:
    layout_data = json.load(f)

output_data = []

for block in layout_data:
    bbox = block.get("bbox")
    block_type = block.get("type", "Text")

    if not bbox or len(bbox) != 4:
        continue  # skip invalid bbox

    x1, y1, x2, y2 = map(int, bbox)#get image crop cordinates
    cropped = image.crop((x1, y1, x2, y2))

    # Run OCR
    text = pytesseract.image_to_string(cropped, lang='eng+hin+ara')
    lang = detect(text) if text else "unknown"#detect language

    output_data.append({
        "bbox": bbox,
        "type": block_type,
        "text": text,
        "language": lang
    })

with open(output_json_path, "w", encoding="utf-8") as f:
    json.dump(output_data, f, ensure_ascii=False, indent=2)

print("Output saved to:", output_json_path)
```
